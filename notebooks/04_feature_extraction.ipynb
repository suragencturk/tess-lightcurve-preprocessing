```python
import numpy as np
import pandas as pd
import lightkurve as lk
import matplotlib.pyplot as plt
```

```python
def load_tess_lightcurve(tic_id):
    search = lk.search_lightcurve(
        f"TIC {tic_id}",
        mission="TESS",
        author="SPOC"
    )
    lc = search.download().PDCSAP_FLUX
    return lc
    lc

```

```python
tic_id = 25155310
lc = load_tess_lightcurve(tic_id)
lc

```

```python
lc = lc.remove_nans().remove_outliers(sigma=5)
```

```python
periods = np.linspace(0.5, 20, 5000)  # gün cinsinden
bls = lc.to_periodogram(
    method="bls",
    period=periods,
    duration=0.1
)
```

```python
bls.plot()
plt.show()
```

```python
best_period = bls.period_at_max_power
best_period
```

```python
folded_lc = lc.fold(period=best_period)
folded_lc.plot()
plt.show()
```
