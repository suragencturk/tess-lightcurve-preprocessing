```python
import lightkurve as lk
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_tess_lightcurve(tic_id):
    search = lk.search_lightcurve(
        f"TIC {tic_id}",
        mission="TESS",
        author="SPOC"
    )
    lc = search.download().PDCSAP_FLUX
    return lc
```

```python
lc = load_tess_lightcurve(25155310)
lc.plot()
```

```python
lc_clean = lc.remove_nans().remove_outliers(sigma=5)
lc_clean
```

```python
lc_flat = lc_clean.flatten(window_length=401)
lc_flat
```

```python
plt.figure(figsize=(10,4))
plt.plot(lc_flat.time.value, lc_flat.flux, ".", markersize=1)
plt.xlabel("Time [BTJD]")
plt.ylabel("Normalized Flux")
plt.title("Preprocessed TESS Light Curve")
plt.show()
```
