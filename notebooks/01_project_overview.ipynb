# TESS False Positive Transit Classification

## Project Motivation
The discovery of exoplanets via transit photometry is strongly affected by
false positive signals such as instrumental noise, stellar variability, and
non-planetary eclipsing systems.

This project aims to develop a machine learning pipeline that can distinguish
true exoplanet transit signals from false positives using TESS light curve data.

## Objectives
- Load and preprocess TESS light curves for multiple targets
- Extract physically meaningful features from transit signals
- Train a machine learning classifier to identify false positives
- Build a reusable pipeline applicable to multiple TESS objects

## Scientific Relevance
This project is designed as a research-oriented mini project suitable for
applications to astronomy research internships (e.g., MPIA).
