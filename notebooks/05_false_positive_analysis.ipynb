```python
import lightkurve as lk
import numpy as np
import matplotlib.pyplot as plt

```

```python
def load_tess_lightcurve(tic_id):
    search = lk.search_lightcurve(
        f"TIC {tic_id}",
        mission="TESS",
        author="SPOC"
    )
    lc = search.download().PDCSAP_FLUX
    return lc

```

```python
tic_id = 25155310
lc = load_tess_lightcurve(tic_id)
lc
```

```python
lc = lc.remove_nans().remove_outliers(sigma=5)
```

```python
periods = np.linspace(0.5, 20, 3000)

bls = lc.to_periodogram(
    method="bls",
    period=periods,
    duration=0.1
)

best_period = bls.period_at_max_power
best_period
```

```python
folded = lc.fold(period=best_period)
folded.plot()
plt.show()
```

```python
oot = folded[(folded.phase < -0.2) | (folded.phase > 0.2)]

oot_std = np.std(oot.flux)
oot_std
```

# --- CONCLUSION ---
# High out-of-transit variability detected (OOT std ~ 17)
# Candidate classified as false positive
